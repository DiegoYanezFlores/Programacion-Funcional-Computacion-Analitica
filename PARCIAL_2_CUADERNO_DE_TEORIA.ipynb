## 1. El Problema del Mundo Real en IA
Cuando entrenamos modelos de Machine Learning con millones de datos, los bucles tradicionales (`for`, `while`) y las variables que cambian constantemente (mutables) generan cuellos de botella en la memoria y errores al intentar procesar datos en paralelo.

La **Programación Funcional** soluciona esto usando matemáticas: funciones que no alteran los datos originales y que siempre dan el mismo resultado.

## 2. Funciones Puras e Inmutabilidad
Una **función pura** tiene dos reglas inquebrantables:
1. **Misma entrada = Misma salida:** Si le pasas el número 2, siempre devolverá el mismo resultado, sin depender de variables externas.
2. **Cero Efectos Secundarios (Inmutabilidad):** No modifica variables globales, ni altera la lista o el dato original que se le pasó.

```python
# ==========================================
# EJEMPLO 1: EL PROBLEMA (Código Impuro y Mutante)
# ==========================================
precios_crudos = [100.0, 200.0, 300.0]

def aplicar_impuesto_impuro(lista_precios):
    for i in range(len(lista_precios)):
        lista_precios[i] = lista_precios[i] * 1.15 # ¡Peligro! Estamos destruyendo la data original
    return lista_precios

print("--- Paradigma Imperativo ---")
print(f"Resultado: {aplicar_impuesto_impuro(precios_crudos)}")
print(f"Data Original destruida: {precios_crudos}\n")
```

```python
# ==========================================
# EJEMPLO 2: LA SOLUCIÓN (Código Puro)
# ==========================================
precios_crudos_2 = [100.0, 200.0, 300.0]

def aplicar_impuesto_puro(precio):
    return precio * 1.15 # Crea un valor nuevo, no toca nada externo

print("--- Paradigma Funcional ---")
# Aquí vemos que la función hace el cálculo correctamente...
print(f"Cálculo puro de 100: {aplicar_impuesto_puro(100.0)}")
# ...y nuestra data original sigue intacta (Inmutabilidad)
print(f"Data Original intacta: {precios_crudos_2}")
```

## 3. Funciones de Primera Clase, Map y Filter
En Python, las funciones pueden guardarse en variables y pasarse como parámetros a otras funciones (a esto se le llama *Funciones de Orden Superior*).

Para aplicar nuestra función pura a una lista gigante sin usar el lento bucle `for`, usamos dos herramientas optimizadas:
* **`map(funcion, datos)`:** Transforma cada elemento aplicando la función.
* **`filter(funcion_condicion, datos)`:** Deja pasar solo los elementos que cumplen la condición.

```python
# Tenemos un pequeño dataset simulado de sensores
datos_sensores = [100, 50, 200, 10, 300]

def aplicar_aumento(datos_sensores_aumentados):
  return datos_sensores_aumentados * 2

# 1. USANDO MAP (Transformación)
# Aplicamos la función pura que creamos arriba a toda la lista
datos_procesados = list(map(aplicar_aumento, datos_sensores))

print("1. Uso de MAP:")
print(f"Originales:  {datos_sensores}")
print(f"Procesados:  {datos_procesados}\n")

# 2. USANDO FILTER (Selección)
# Creamos una función pura que devuelve True o False
def es_anomalia(valor):
    return valor > 150

datos_filtrados = list(filter(es_anomalia, datos_sensores))

print("2. Uso de FILTER:")
print(f"Anomalías detectadas (>150): {datos_filtrados}")
```

## 🎯 RETO PRÁCTICO: Refactorización (Data Pipeline)
**Contexto:** Estás limpiando un dataset de temperaturas para entrenar una Red Neuronal. El ingeniero anterior dejó un "código espagueti" imperativo.

**Tu Misión:**
1. Reescribe la limpieza de datos usando **SOLO** funciones puras, `map()` y `filter()`.
2. 🚫 **Regla estricta:** No puedes usar `for`, `while`, ni `.append()`.

```python
# DATASET ORIGINAL (No lo modifiques)
temperaturas_crudas = [22.5, "error_sensor", 25.1, 19.8, "fail", 30.2, 15.0]

# ==========================================
# EL CÓDIGO ESPAGUETI (Lo que debes mejorar)
# ==========================================
resultados_imperativos = []
for temp in temperaturas_crudas:
    if type(temp) == float: # Filtrar textos
        f_temp = (temp * 9/5) + 32 # Convertir C a F
        if f_temp > 70: # Quedarse con > 70
            resultados_imperativos.append(f_temp)

print(f"Resultado esperado: {resultados_imperativos}")

# ==========================================
# TU SOLUCIÓN FUNCIONAL (Escribe tu código aquí abajo)
# ==========================================
# 1. Crea una función para validar si es float
# 2. Crea una función para convertir a Fahrenheit
# 3. Crea una función para evaluar si es mayor a 70
# 4. Aplica filter y map para armar tu pipeline

def es_numero(valor):
  return isinstance(valor, (float))

def convertir_a_fahrenheit(valor):
  return (valor * 9/5) + 32

def es_mayor_a_70(valor):
  return valor > 70

solo_numeros = (filter(es_numero, temperaturas_crudas))
convertir_a_fahrenheit = (map(convertir_a_fahrenheit, solo_numeros))
resultado_funcional = (filter(es_mayor_a_70, convertir_a_fahrenheit))

print("Resultado funcional: ", list(resultado_funcional))

#SOLUCION "ONE-LINER"

resultado_final = list(filter(lambda t : t > 70, map(lambda tc : (tc * 9/5) + 32, filter(lambda t: isinstance(t, (float)), temperaturas_crudas))))

print("Resultado final: ", resultado_final)
```

## Funciones Puras (Pure Functions)
Una función pura es, en esencia, la traducción exacta de una función matemática al código.

Para que una función sea considerada "pura", debe cumplir obligatoriamente con dos reglas de oro:

* Regla 1: Determinismo (Misma entrada = Misma salida).

Si le pasas el número 5 a una función pura, y esta devuelve 10, entonces siempre que le pases un 5 debe devolver 10. No importa si es de día, de noche, si la base de datos se cayó o si el usuario hizo clic en otro lado. Su resultado depende única y exclusivamente de los parámetros que le entregas.

* Regla 2: Cero Efectos Secundarios (No Side Effects).

Una función pura entra, hace su trabajo, devuelve un valor nuevo y se va sin tocar nada del mundo exterior. No modifica variables globales, no altera la lista original que le pasaste, no escribe en un archivo y no hace llamadas a la red. Es completamente inofensiva y aislada.



```python
# ❌ FUNCIÓN IMPURA (Rompe las dos reglas)
impuesto = 0.15 # Variable externa
carrito = [100, 200]

def aplicar_impuesto_impuro():
    global impuesto
    # Efecto secundario: modifica una variable externa directamente
    carrito.append(carrito[0] * impuesto)
    return carrito

# ✅ FUNCIÓN PURA (Cumple las dos reglas)
def aplicar_impuesto_puro(precio, tasa_impuesto):
    # Depende solo de sus parámetros y retorna un valor nuevo
    # No modifica nada de afuera.
    return precio + (precio * tasa_impuesto)
```

## Funciones de Orden Superior (Higher-Order Functions)

En muchos lenguajes tradicionales, las funciones y las variables viven en mundos separados. En la programación funcional (y en lenguajes como Python o JavaScript), las funciones son "ciudadanos de primera clase". Esto significa que una función es un dato más, igual que un número o un texto. Puedes guardar una función en una variable.

A partir de esto, nace la Función de Orden Superior, que es simplemente una función que hace al menos una de estas dos cosas:

1. Recibe una o más funciones como parámetros (como argumentos).

2. Devuelve una función como resultado.

Piensa en una función de orden superior como un "gerente". El gerente no hace el trabajo manual, sino que tú le entregas a un "empleado" (otra función) y una "lista de tareas" (datos), y el gerente se encarga de aplicar ese empleado a todas las tareas.

Ejemplo con los "gerentes" más famosos (map y filter):

Imagina que tienes una lista de datos y una función pura. En lugar de hacer un bucle for (que es imperativo), usas una función de orden superior para conectar ambas cosas.

```python
datos = [1, 2, 3, 4, 5]

# Esta es nuestra función pura ("el empleado")
def multiplicar_por_dos(numero):
    return numero * 2

# 'map' es la función de orden superior ("el gerente").
# Le pasamos la función como si fuera una variable (sin los paréntesis).
resultado = list(map(multiplicar_por_dos, datos))

print(resultado)
# Salida: [2, 4, 6, 8, 10]
```

## 1. Expresiones Lambda: Funciones "Desechables"
En la clase anterior aprendimos a usar mas a profundidad `map` y `filter` pasando funciones creadas con `def`. Sin embargo, a veces necesitamos una función matemática muy simple que solo usaremos una vez. No vale la pena gastar líneas de código nombrándola y documentándola.

Para eso existen las **Expresiones Lambda** (o funciones anónimas). Son funciones de una sola línea, puras por naturaleza, que se crean y se destruyen en el acto.

**Sintaxis:** `lambda parametros: expresion_a_retornar`

```python
# ==========================================
# 1. LAMBDAS: LA FORMA ELEGANTE
# ==========================================
dataset_precios = [15.50, 45.00, 120.99, 5.00, 300.50]

print("--- Data Original ---")
print(dataset_precios)

# FORMA TRADICIONAL (Muchas líneas para algo simple)
def aplicar_descuento(precio):
    return precio * 0.8

precios_desc_tradicional = list(map(aplicar_descuento, dataset_precios))

# FORMA LAMBDA (Directo al punto, código limpio)
precios_desc_lambda = list(map(lambda p: p * 0.8, dataset_precios))

# Filtrar solo los precios caros (>100) usando lambda
precios_caros = list(filter(lambda p: p > 100, dataset_precios))

print("\n--- Resultados Lambda ---")
print(f"Con descuento: {precios_desc_lambda}")
print(f"Solo caros: {precios_caros}")
```

## 2. El Problema del Mundo Real: Out of Memory (OOM) 💥
Imagina que trabajas en OpenAI o Google y debes entrenar un modelo de lenguaje con **100 Gigabytes** de texto. Tu computadora (o el servidor que alquilaste) solo tiene **16 GB de Memoria RAM**.

Si intentas cargar todo el texto en una variable de Python tipo lista (ej. `textos = leer_archivo_100gb()`), la RAM se llenará al 100%, la computadora se congelará y el programa lanzará el temido error: `MemoryError` (OOM).

## 3. La Solución: Generadores y Evaluación Perezosa (Lazy Evaluation)
La programación funcional nos da una herramienta mágica: **Los Generadores**.

En lugar de cargar los 100 millones de registros en la memoria al mismo tiempo, un generador te entrega **un solo registro a la vez**, y luego se pausa. Cuando el modelo de IA procesa ese dato, el generador avanza al siguiente. A esto se le llama "Evaluación Perezosa" porque solo calcula el dato en el momento exacto en que se lo pides, ni un segundo antes.

## 4. La Magia del "Yield" (Evaluación Perezosa)
El problema: No entienden cómo funciona yield y por qué es diferente a return.
La analogía: El chef de hamburguesas y el botón de pausa.

Piensen en una función tradicional con return como un chef de comida rápida al que le piden 100 hamburguesas. El chef cierra la puerta de la cocina, hace las 100 hamburguesas, las pone TODAS en una bandeja gigante y te las entrega de golpe. (El return termina el trabajo y te da todo).

Una función con yield (un generador) es como un chef personal. Le pides 100 hamburguesas, pero él hace SOLO UNA, te la entrega y se queda CONGELADO (en pausa) mirándote.

No hará la número 2 hasta que tú te comas la primera y le digas: 'Dame la siguiente'. (En Python, esa orden es la función next()).

El paso a paso en el código:

Cuando Python lee la palabra yield, hace exactamente tres cosas:

1. Entrega el dato actual al modelo de IA.

2. Pone "Pausa" a la función (como ponerle pausa a una película). Recuerda exactamente en qué línea de código se quedó y qué valores tenían las variables.

3. Espera pacientemente. No consume procesador ni llena la memoria. Solo cuando el modelo de IA dice "dame otro dato", la función quita la pausa, da un paso más en el bucle, y hace el siguiente yield.

A esto se le llama Evaluación Perezosa (Lazy Evaluation) porque la computadora es literalmente "perezosa": no va a calcular el millón de datos si tú solo le has pedido el primero.

```python
# Simulador de lectura de un archivo inmenso de registros médicos
def lector_registros_masivos_mock():
    registros = [
        "Paciente_1,Saludable,120bpm",
        "Paciente_2,Anomalía,150bpm",
        "Paciente_3,Saludable,110bpm"
    ]
    # En la vida real, aquí no habría una lista, sino un for abriendo un archivo de 50GB línea por línea.
    print("Iniciando conexión con base de datos masiva...")
    for registro in registros:
        print(f"[Log del Sistema: Procesando la fila en memoria...]")
        yield registro # Entregamos la fila y nos pausamos!

# Instanciamos el generador
pipeline_datos = lector_registros_masivos_mock()

print("--- Iniciando Entrenamiento del Modelo ---")
# El modelo le pide UN dato al generador
dato_1 = next(pipeline_datos)
print(f"Modelo IA Entrenando con: {dato_1}\n")

# El modelo le pide el SEGUNDO dato
dato_2 = next(pipeline_datos)
print(f"Modelo IA Entrenando con: {dato_2}\n")

dato_3 = next(pipeline_datos)
print(f"Modelo IA Entrenando con: {dato_3}\n")
```

## 5. El "Paracaídas de Emergencia": Manejo de Errores con Try/Except


En el mundo de la Inteligencia Artificial, los datos nunca son perfectos. Si estamos procesando 10 millones de registros de un sensor, es casi seguro que algunos vendrán con fallas, vacíos o textos sin sentido (ej. `"error_conexion"` en lugar de un número).

**El Problema:** Por defecto, si Python intenta hacer una operación matemática con un texto, entra en pánico, lanza un texto rojo gigante (una Excepción) y **destruye o apaga todo el programa**. Si llevabas 9 horas entrenando tu modelo y en la hora 10 llega un dato corrupto, lo pierdes todo.

**La Solución: Try / Except**
Es el equivalente a un paracaídas de emergencia o un plan de contingencia. Le decimos a la computadora:
> *"INTENTA (`try`) ejecutar este código. Si algo explota o sale mal, por favor no te mueras ni te cierres. En su lugar, atrapa el error y ejecuta esta EXCEPCIÓN (`except`) como Plan B, y luego sigue trabajando con el resto de los datos."*

```python
# ==========================================
# CÓMO FUNCIONA EL TRY / EXCEPT
# ==========================================

dato_corrupto = "falla_de_sensor"

print("--- Sin Try/Except (Lo que pasaría normalmente) ---")
# Si descomentas la línea de abajo, el programa morirá inmediatamente
# numero = float(dato_corrupto)
print("Si corres la línea de arriba, este mensaje NUNCA se imprimiría porque el programa ya murió.\n")


print("--- Con Try/Except (El programa sobrevive) ---")
try:
    # 1. El Camino Feliz (Lo que queremos que pase)
    print(f"Intentando convertir '{dato_corrupto}' a decimal...")
    numero = float(dato_corrupto)
    print(f"Éxito. El número es {numero}")

except ValueError:
    # 2. El Plan B (Lo que pasa si falla el Camino Feliz)
    # Atrapamos específicamente el error "ValueError" (Error de valor)
    print("¡Ups! Hubo un error, el dato no era un número válido.")
    print("Pero el programa no se apagó. Podemos asignar un valor por defecto (ej. 0.0) o ignorarlo.")
    numero = 0.0

print(f"\nEl programa continuó sin problemas. Valor final guardado: {numero}")
```

### ¿Cómo se usa en nuestros Generadores (Pipelines)?
Cuando procesamos datos masivos con la Evaluación Perezosa (`yield`), usamos `try/except` dentro del bucle.

Si el dato es bueno, lo procesamos y hacemos `yield`. Si el dato es malo, el `except` atrapa el error y simplemente usamos la palabra reservada **`continue`** para decirle a Python: *"Ignora este dato y salta inmediatamente al siguiente"*. ¡Así nuestro pipeline nunca se detiene!

## 6. El "Data Pipeline": Uniendo todas las piezas 🏭


Imagina una fábrica de jugos. Entran camiones con manzanas recién cosechadas (Datos Crudos). Algunas vienen podridas o traen piedras por accidente (Errores).

Nuestro **Pipeline (Tubería)** tendrá 3 estaciones de trabajo:
1. **El Inspector (Generador + Try/Except):** Toma una manzana, si es una piedra la tira sin detener la fábrica (Manejo de errores), si es buena, la deja pasar (`yield`).
2. **El Pelador (Map):** A cada manzana que le llega, le quita la cáscara y la corta.
3. **El Control de Calidad (Filter):** Solo deja pasar al envase final los trozos que pesan más de cierto límite.

En Inteligencia Artificial hacemos exactamente lo mismo con los datos. Vamos a construir un Pipeline para un sensor de temperatura:

```python
# 0. LOS DATOS CRUDOS (El camión de manzanas)
# Vienen textos, números y un error de red mezclados.
lecturas_sensor = ["20", "30", "ERROR_WIFI", "40", "10"]

print(f"Datos originales: {lecturas_sensor}\n")

# ==========================================
# ESTACIÓN 1: El Inspector (Generador con Try/Except)
# ==========================================
def limpiar_textos(datos):
    for dato in datos:
        try:
            # Intenta convertir el texto a decimal
            numero = float(dato)
            yield numero # Si funciona, lo deja pasar a la siguiente estación
        except ValueError:
            # Si da error (ej. "ERROR_WIFI"), lo ignora y la fábrica sigue funcionando
            print(f"⚠️ Alerta: Se detectó y descartó basura ('{dato}')")
            continue

# ==========================================
# CONSTRUYENDO LA TUBERÍA (PIPELINE)
# ==========================================
# Aquí conectamos las estaciones. ¡Recuerda que los generadores son perezosos!
# Hasta que no encendamos la fábrica al final, no se procesará nada.

# Paso 1: Pasan por el inspector (salen como números: 20.0, 30.0, 40.0...)
datos_limpios = limpiar_textos(lecturas_sensor)

# Paso 2: El Pelador (MAP) -> Les sumamos 5 grados de calibración a todos
datos_calibrados = map(lambda temp: temp + 5, datos_limpios)

# Paso 3: Control de Calidad (FILTER) -> Solo dejamos pasar las alertas (> 30 grados)
alertas_finales = filter(lambda temp: temp > 30, datos_calibrados)

# ==========================================
# ENCENDIENDO LA FÁBRICA
# ==========================================
print("\nIniciando procesamiento de la tubería...")

# Al usar list() obligamos a la tubería a encenderse y jalar todos los datos
resultado = list(alertas_finales)

print(f"\n✅ Temperaturas peligrosas detectadas: {resultado}")
# Resultado esperado: [35.0, 45.0]
```

### 🧠 ¿Por qué este Pipeline es nivel Profesional (Senior)?
1. **Es Seguro:** Gracias al `try/except`, el texto `"ERROR_WIFI"` no hizo que el programa explotara.
2. **Es Ligero en Memoria:** Gracias al `yield` (Generador), si en lugar de 5 datos tuviéramos 5 millones, nuestra computadora usaría la misma cantidad mínima de memoria RAM, porque los datos pasan de uno en uno.
3. **Es Matemáticamente Puro:** Usamos `map` y `filter` con funciones `lambda`. No modificamos la lista original, simplemente transformamos los datos a medida que fluyen por la tubería.

Antes de pasar a la documentación, debemos conocer una última herramienta de la Unidad 2. Ya sabemos usar `map` y `filter`, pero Python tiene una sintaxis nativa muy elegante llamada **List Comprehension** (Comprensión de Listas) que nos permite hacer ambas cosas en una sola línea legible.

```python
# ==========================================
# LIST COMPREHENSIONS: Elegancia en 1 línea
# ==========================================
temperaturas = [15.0, 30.5, 40.2, 10.0, 50.1]

# Queremos: Filtrar las mayores a 20 y sumarles 5 grados.

# Opción A: Usando map y filter (Funcional puro)

opcion_a = list(map(lambda x: x + 5, filter(lambda x: x > 20, temperaturas)))

# Opción B: Usando List Comprehension (Estilo Pythonico)
# Sintaxis: [operacion for variable in lista if condicion]

opcion_b = [temp + 5 for temp in temperaturas if temp > 20]

print(f"Resultado Map/Filter: {opcion_a}")
print(f"Resultado List Comp: {opcion_b}")

# Nota de IA: Las List Comprehensions son geniales y rápidas para pocos datos,
# pero recuerda usar Generadores (yield) si los datos son masivos para no saturar la RAM.

```

## 5. Type Hinting (Anotaciones de Tipo)
A diferencia de Java, Python es de **tipado dinámico**. Puedes meter un texto en una variable que tenía un número y Python no se quejará... hasta que el programa explote en producción.

En proyectos de IA, donde procesamos matrices matemáticas, no saber qué tipo de dato recibe una función es un suicidio técnico. Para arreglar esto, usamos **Type Hints**. No cambian cómo funciona el código, pero le dicen a tu equipo: *"Esta función DEBE recibir una Lista de decimales, y DEBE retornar un solo decimal"*.

```python
from typing import List

# Función CON TIPADO (Transparente y Segura)
# Le indicamos explícitamente que recibe dos listas de floats, y que retorna un float
def calcular_error_medio_seguro(predicciones: List[float], reales: List[float]) -> float:
    return sum([abs(p - r) for p, r in zip(predicciones, reales)]) / len(predicciones)

valores_predichos = [1.2, 3.4, 5.0]
valores_reales = [1.0, 3.5, 4.8]

error = calcular_error_medio_seguro(valores_predichos, valores_reales)
print(f"Error Medio Absoluto: {error:.2f}")
```

```python
def normalizar_datos(datos: List[float], valor_maximo: float) -> List[float]:
    """
    Normaliza una lista de datos numéricos a una escala de 0 a 1.

    Paso de preprocesamiento vital para capas de Redes Neuronales.

    Args:
        datos (List[float]): Lista de valores crudos a normalizar.
        valor_maximo (float): El valor teórico más alto posible.

    Returns:
        List[float]: Nueva lista con valores escalados.
    """
    return [x / valor_maximo for x in datos]

# Python nos muestra el manual si usamos help()
help(normalizar_datos)
```

## 1. La Ilusión de la Aleatoriedad
En IA, usamos números aleatorios todo el tiempo (para mezclar datos, inicializar los "pesos" de una Red Neuronal, etc.). Pero las computadoras no son realmente aleatorias; usan algoritmos matemáticos complejos para *simular* aleatoriedad (Pseudo-aleatoriedad).

Ejecuta la celda de abajo **varias veces**. Notarás que el resultado cambia en cada ejecución.

```python
import random

def inicializar_pesos_neuronales() -> list[float]:
    """Simula la creación de 3 pesos aleatorios iniciales para una IA."""
    return [random.random() for _ in range(3)]

# Ejecuta esta celda 3 veces. ¡Los números cambiarán siempre!
print(f"Pesos de la IA: {inicializar_pesos_neuronales()}")
```

### 🔍 Anatomía del Código:
* `import random`: Traemos la librería nativa de Python encargada de la aleatoriedad.
* `-> list[float]`: Usamos Type Hinting moderno para indicar que la función devolverá una lista de números decimales.
* `for _ in range(3)`: Es un bucle que se repite 3 veces. Usamos el guion bajo `_` como nombre de variable porque es una convención profesional en Python que significa: *"Necesito que el bucle gire, pero no me importa usar el número de la vuelta"*.
* `random.random()`: Genera un número decimal (float) entre 0.0 y 1.0.

## 2. Domando el Caos: La Semilla (Seed)
Si los pesos cambian cada vez que ejecutas el código, tu IA aprenderá de forma diferente en cada intento. Para evitar esto, usamos una **Semilla Aleatoria (`seed`)**.

La semilla es simplemente un número entero (el más famoso en la comunidad de programación es el `42`). Al fijar una semilla, le decimos a la computadora: *"Quiero que me des números aparentemente aleatorios, pero si te doy la misma semilla mañana, dame exactamente la misma secuencia"*.

```python
import random

# ==========================================
# EXPERIMENTO REPRODUCIBLE
# ==========================================
# Al fijar la semilla, congelamos el universo matemático.
SEMILLA_UNIVERSAL = 42
random.seed(SEMILLA_UNIVERSAL)

"""Simula la creación de 3 pesos aleatorios iniciales para una IA."""
def inicializar_pesos_seguros() -> list[float]:
    return [random.random() for _ in range(3)]

# No importa cuántas veces ejecutes TODO este bloque de código,
# el resultado SIEMPRE será exactamente el mismo.
print(f"Pesos Congelados: {inicializar_pesos_seguros()}")
```

### 🔍 Anatomía del Código:
* `SEMILLA_UNIVERSAL = 42`: Creamos una constante (por eso va en mayúsculas según PEP 8). El número 42 es una tradición entre programadores (referencia al libro *Guía del autoestopista galáctico*), pero puedes usar tu año de nacimiento o cualquier número entero.
* `random.seed(SEMILLA_UNIVERSAL)`: **¡La línea más importante!** Esto congela el motor matemático de Python. A partir de este momento, los números "aleatorios" seguirán un guion predecible. Si cierras Colab, vuelves mañana y ejecutas esto, te dará exactamente la misma secuencia de números.

```python
# Datos de prueba
precio_arranque = 100.0
dias = 5
semilla_bot = 99
# def simular_precios(...):
#     pass
```

```python
import random
from typing import Iterator

def simular_precios(precio_inicial: float, cantidad_dias: int, semilla: int) -> Iterator[float]:
    """
    Simula la fluctuación diaria del precio de un activo usando un generador reproducible.

    Args:
        precio_inicial (float): Valor de arranque del activo.
        cantidad_dias (int): Número de días (iteraciones) a simular.
        semilla (int): Semilla para garantizar que la fluctuación sea idéntica en futuras pruebas.

    Yields:
        float: Precio resultante al finalizar cada día simulado.
    """
    random.seed(semilla)
    precio_actual = precio_inicial

    for _ in range(cantidad_dias):
        # Fluctuación aleatoria entre -$5.0 y +$5.0
        cambio_diario = random.uniform(-5.0, 5.0)
        precio_actual += cambio_diario
        yield round(precio_actual, 2)

# Prueba del Docente:
simulacion = simular_precios(100.0, 5, 300)
print("Simulación de precios (Semilla 99):")
# Demostramos que el generador funciona iterándolo de forma funcional
for dia, precio in enumerate(simulacion, 1):
    print(f"Día {dia}: ${precio}")

# Resultado esperado SIEMPRE:
# Día 1: $101.76, Día 2: $99.85, Día 3: $100.2, Día 4: $101.44, Día 5: $100.51
```

* **`precio_actual = precio_inicial`**: Creamos una variable local para ir llevando la cuenta del precio. Nunca debemos modificar directamente el parámetro `precio_inicial` (esto evita efectos secundarios indeseados, respetando la Programación Funcional).
* **`for _ in range(cantidad_dias):`**: Usamos el guion bajo `_` porque necesitamos que el bucle dé un número exacto de vueltas (los días), pero no nos interesa usar el número de la vuelta actual (el día 1, 2, 3...) dentro de la matemática.
* **`random.uniform(-5.0, 5.0)`**: A diferencia de `randint` que solo da números enteros, `uniform` nos devuelve un número decimal (float) continuo entre ese rango. Perfecto para simular centavos en finanzas.
* **`yield round(precio_actual, 2)`**: La magia de la resiliencia de memoria. La función calcula el precio del Día 1, lo redondea a 2 decimales y lo "escupe" (`yield`). Luego **se pausa** y se queda congelada esperando a que le pidan el Día 2. Nunca construye una lista gigante en la RAM.

```python
# ==========================================
# SOLUCIÓN MAESTRA Y ARQUITECTURA PROFESIONAL
# ==========================================

import random
from typing import List, Iterator

# Los datos de prueba
datos_crudos = ["22.5", "error_conexion", "25.1", "29.9", "sensor_roto", "18.5", "30.2"]

def limpiar_temperaturas(datos: List[str], semilla: int) -> Iterator[float]:
    """
    Generador que procesa lecturas de sensores, filtrando errores y simulando ruido termico reproducible

    Args:
        datos(List[str]): Lista de lecturas en formato de texto.
        semilla (int): Semilla para reproducibilidad, Numero para fijar el motor de aleatoriedad

    Yields:
        float: Temperatura numerica validad y ajustada (Fase 'Transform')
    """
    random.seed(semilla)

    for dato in datos:
      try:
          # Si el dato es texto basura ("Error de conexion"), float() fallara y se saltara al except
          temp_base = float(dato)
          ruido = random.uniform(0.0, 0.5)
          yield round(temp_base + ruido, 2)
      except ValueError:
          # Ignorar la basura silenciosamente protengiendo el pipeline
          continue

def detectar_alertas_termicas(temperaturas: Iterator[float], limite_critico: float) -> List[float]:
    """
    Filtra un flujo de temperaturas devolviendo solo aquellas que superan un límite crítico.

    Args:
        temperaturas (Iterator[float]): Flujo de datos térmicos (generador).
        limite_critico (float): Umbral de temperatura máxima permitida.

    Returns:
        List[float]: Lista de temperaturas que representan una alerta.
    """
    # Enfoque Funcional Puro: List Comprehension sin variables globales
    return [temp for temp in temperaturas if temp > limite_critico]


# ==========================================
# EJECUCIÓN DEL PIPELINE (Fase 'Load')
# ==========================================
print("--- INICIANDO SISTEMA DE AUDITORÍA TÉRMICA ---")

# 1. Instanciamos el generador (Aún no procesa nada, ahorra memoria RAM)
generador_datos = limpiar_temperaturas(datos_crudos, semilla=101)

# 2. Consumimos el generador evaluando las reglas de negocio
alertas_oficiales = detectar_alertas_termicas(generador_datos, limite_critico=28.0)

print(f"Temperaturas Críticas Detectadas: {alertas_oficiales}")
```


## Inmutabilidad y Diseño Defensivo

Para resolver los siguientes retos como un profesional, necesitamos pulir dos conceptos de ingeniería que separan el código "que funciona" del código "a prueba de balas".

### 1. El Principio de Inmutabilidad


En el código tradicional, solemos usar listas y modificarlas constantemente usando `.append()`, `.pop()`, o reasignando valores. A esto se le llama **mutabilidad**. En proyectos pequeños, está bien. Pero en Inteligencia Artificial, donde procesamos datos en paralelo (usando múltiples núcleos del procesador al mismo tiempo), la mutabilidad es una pesadilla: si dos procesos intentan modificar la misma lista al mismo tiempo, el programa colapsa.

**La solución funcional es la Inmutabilidad:** En lugar de modificar los datos originales, creamos una copia transformada.
* **Malo (Mutando estado global):** `mi_lista.append(nuevo_dato)`
* **Bueno (Inmutable y puro):** `return [dato for dato in datos_viejos]` -> Entregas una lista completamente nueva. Los datos originales quedan intactos.

### 2. Diseño Defensivo (Fail-Safe) en Pipelines
Un *Data Pipeline* es como una línea de ensamblaje de fábrica. Si entra una pieza defectuosa y la máquina no sabe qué hacer, toda la fábrica se detiene.
En programación, esto ocurre cuando asumimos que los datos siempre vendrán perfectos. El **Diseño Defensivo** asume lo peor: asume que los sensores fallarán, que los usuarios escribirán letras en lugar de números, y que el internet se cortará.

Usar el bloque `try / except` dentro de un generador no es solo "manejo de errores", es un **escudo**. Permite que la pieza defectuosa sea descartada silenciosamente para que el resto de los millones de datos sigan fluyendo sin interrumpir el modelo de IA.

Construir un Data Pipeline que:

3. Devuelva una lista limpia con los montos finales a cobrar.

```python
from typing import List, Iterator

# Datos de entrada del E-commerce
carritos_crudos = ["100.00", "VACIO", "50.50", "NULL", "200.00", "ERROR_500", "15.99"]

# ==========================================
# 🛠️ TU ESPACIO DE TRABAJO
# ==========================================
# 1. Generador de limpieza de carritos

def extraer_precios_validos(carritos: List[str]) -> Iterator[float]:
  """ Generador que ignora carritos vacios o con errores de sistema """
  for carrito in carritos:
    try:
      # Intentamos convertir a decimal. Si es vacio, fallara y salta al except.
      yield float(carrito)
    except ValueError:
      continue

# 2. Aplicación de impuestos (usando map o comprehensions)

def aplicar_impuestos(precios_limpios: Iterator[float]) -> List[float]:
  """ Aplica un 15 % de impusta usnado una List Comprehension (inmutabilidad)"""
  return [round(precio * 1.15, 2) for precio in precios_limpios]

# 3. Imprime el resultado de los cobros finales

generador_ventas = extraer_precios_validos(carritos_crudos)
cobro_final = aplicar_impuestos(generador_ventas)

print(f"Cobros Finales con impuestos: {cobro_final}")
```

## Reto 2 - Detección de Fraude Bancario (Pipelines Complejos)

**Contexto:**
Eres el Arquitecto de Datos de un banco internacional. Se te entrega un registro de transferencias internacionales. Además, todos los montos están en Euros y necesitan ser convertidos a Dólares (Tasa: 1 EUR = 1.10 USD).

**Tu Misión:**
Construye un pipeline de 3 pasos que aplique Inmutabilidad y Diseño Defensivo:
1. **Paso 1 (Extract & Clean):** Un generador que reciba los datos crudos, evada los errores con `try/except` y devuelva los montos válidos en formato `float`.
2. **Paso 2 (Transform - Map):** Una función pura que reciba el iterador de datos limpios y devuelva una lista nueva (inmutable) con los montos convertidos a Dólares (multiplicados por 1.10).
3. **Paso 3 (Transform - Filter):** Una función pura que reciba la lista en dólares y extraiga solo las transacciones sospechosas (mayores a $10,000 USD).

```python
from typing import List, Iterator

# Datos bancarios crudos (Mezcla de montos en Euros y basura del sistema)
transacciones_crudas = ["5000.50", "SYSTEM_FAIL", "12500.00", "850.25", "NULL", "9500.00", "CORRUPTED_DATA"]

# ==========================================
# 🛠️ TU ESPACIO DE TRABAJO
# ==========================================
# 1. Generador de limpieza (Diseño Defensivo)
def limpiar_transacciones(transacciones: List[str]) -> Iterator[float]:
  """Generador que limpia y filtra transacciones."""
  for transaccion in transacciones:
    try:
      yield float (transaccion)
    except ValueError:
      continue


# 2. Función de conversión de moneda (Inmutabilidad, usar map o comprehensions)
def convertir_a_dolares(transacciones_limpias: Iterator[float]) -> List[float]:
  """Función pura que aplica inmutabilidad creando una lista nueva."""
  return [round(monto * 1.10, 2) for monto in transacciones_limpias]


# 3. Función de detección de fraude (Filtro > 10000)
def detectar_fraudes(transacciones_usd: List[float], limite: float = 10000.00) -> List[float]:
  """Función pura que filtra transacciones con monto > 10000.00"""
  return [monto for monto in transacciones_usd if monto > limite]

# 4. Ejecución del Pipeline (Conecta las 3 funciones)
generador_limpio = limpiar_transacciones(transacciones_crudas)
transacciones_dolares = convertir_a_dolares(generador_limpio)
fraudes_encontrados = detectar_fraudes(transacciones_dolares)

print(f"Transacciones con Fraudes: {fraudes_encontrados}")
```
